```python
!apt-get -qq update
!apt-get -qq install -y cmake build-essential git
!git clone -q --depth 1 https://github.com/ggerganov/llama.cpp.git
!cmake -S llama.cpp -B llama.cpp/build -DLLAMA_CURL=OFF
!cmake --build llama.cpp/build --target llama-cli -j 1
```

```python
!pip -q install huggingface_hub
from huggingface_hub import hf_hub_download
model_path=hf_hub_download(repo_id='lmstudio-community/SmolLM2-360M-Instruct-GGUF',filename='SmolLM2-360M-Instruct-Q4_K_M.gguf'); print(model_path)
```

```python
prompt='You are an educational healthcare AI tutor. Explain in simple language what sensitivity means in a diagnostic test. Do not provide medical advice.'
cmd=['./llama.cpp/build/bin/llama-cli','-m',model_path,'-p',prompt,'-c','512','-n','120','--temp','0.2','--top-k','20']
r=subprocess.run(cmd,capture_output=True,text=True); print(r.stdout[-4000:]); gc.collect()
```
