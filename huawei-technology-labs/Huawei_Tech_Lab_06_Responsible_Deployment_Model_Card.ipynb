# Huawei Technology Lab 6
## Responsible Deployment with MindSpore + ModelArts

This lab closes the Huawei technology pathway by teaching what must be
documented before a trained model is even considered for deployment.

**Educational prototype — not for clinical diagnosis or treatment.**

## Learning objectives

Learners will:
1. save a MindSpore checkpoint;
2. document preprocessing and evaluation;
3. define intended and prohibited use;
4. create a model card;
5. distinguish an educational web demo from a validated clinical system.

````python
import json
from pathlib import Path
from datetime import datetime

output_dir = Path("responsible_deployment_outputs")
output_dir.mkdir(exist_ok=True)

model_card = {
    "model_name": "Healthcare AI Teaching Demonstration",
    "framework": "MindSpore",
    "cloud_workflow": "Huawei Cloud ModelArts-ready",
    "version": "0.1-educational",
    "date": datetime.now().date().isoformat(),
    "intended_use": ["AI education", "demonstrating model evaluation", "teaching responsible AI documentation"],
    "prohibited_use": ["clinical diagnosis", "screening", "triage", "treatment recommendation", "patient management"],
    "data": {"type": "synthetic or approved public teaching data", "identifiable_patient_data_allowed": False},
    "preprocessing": ["document resizing or feature scaling", "fit preprocessing on training data only", "preserve the test set for final evaluation"],
    "required_evaluation": ["task-appropriate performance metrics", "error analysis", "subgroup assessment where relevant", "calibration where probabilities are interpreted", "external validation before any real clinical claim"],
    "limitations": ["educational dataset", "not externally validated", "not a medical device", "not approved for clinical use"],
}

model_card_path = output_dir / "MODEL_CARD.json"
model_card_path.write_text(json.dumps(model_card, indent=2), encoding="utf-8")
print(model_card_path.read_text(encoding="utf-8"))
````

## Optional: save a MindSpore checkpoint

If a trained MindSpore network named `net` already exists, use:

```python
import mindspore as ms
ms.save_checkpoint(net, "responsible_deployment_outputs/model.ckpt")
```

A checkpoint alone is never enough. It must travel with preprocessing
instructions, model version, evaluation results, and the model card.

````python
deployment_manifest = {
    "status": "educational_demo_ready",
    "clinical_deployment": False,
    "modelarts_deployment_executed": False,
    "required_before_real_use": ["institutional governance", "privacy/security review", "representative external validation", "clinical workflow evaluation", "regulatory assessment where applicable"],
}

manifest_path = output_dir / "DEPLOYMENT_MANIFEST.json"
manifest_path.write_text(json.dumps(deployment_manifest, indent=2), encoding="utf-8")
print(manifest_path.read_text(encoding="utf-8"))
````
