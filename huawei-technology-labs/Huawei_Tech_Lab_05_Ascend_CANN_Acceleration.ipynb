## Build the same MindSpore network regardless of accelerator

```python
from mindspore import nn

class AscendDemoNet(nn.Cell):
    def __init__(self):
        super().__init__()
        self.net = nn.SequentialCell(
            nn.Dense(16, 32),
            nn.ReLU(),
            nn.Dense(32, 2),
        )

    def construct(self, x):
        return self.net(x)

network = AscendDemoNet()
print(network)
```
